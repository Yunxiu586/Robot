A data container stores multiple objects. Element requirements depend on the container type.

## Lists
`[element1, element2, element3, ...]`

```python
import sys

list_result = [x for x in range(1, 100001) if x % 2 == 0]
gen_result = (x for x in range(1, 100001) if x % 2 == 0)
print(f"List comprehension size: {sys.getsizeof(list_result)} bytes")
print(f"Generator expression size: {sys.getsizeof(gen_result)} bytes")
'''
A list comprehension constructs a list from an expression and its for/if clauses.
A generator expression returns an iterator that produces values on demand.
sys.getsizeof() reports memory directly attributed to an object, not the objects it refers to.
''';
```

```python
# Find words that begin with 'b', convert them to uppercase, and keep the other words unchanged
words = ['apple', 'banana', 'cherry', 'date']

# List comprehension with a conditional expression
words_upper = [word.upper() if word.startswith('b') else word for word in words]
'''
[expression_if_true if condition else expression_if_false for variable in iterable]
Every element is processed.
''';
print(words_upper)
```

```python
nums = [1, 2, 3, 4]
evens = [num for num in nums if num % 2 == 0]
'''
[expression for variable in iterable if condition]
Elements that do not satisfy the condition are filtered out.
''';
print(evens)
```

```python
# map()
# Return an iterator that applies a function to each item of an iterable
nums = [1, 2, 3, 4]
squared = list(map(lambda x: x**2, nums))
print(squared)
```

```python
# filter()
# Return an iterator containing the items for which the function is true
nums = [1, 2, 3, 4]
evens = list(filter(lambda x: x % 2 == 0, nums))
print(evens)
```

```python
from functools import reduce
# reduce()
# Apply a two-argument function cumulatively to the items from left to right
nums = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, nums)
print(product)  # 1 * 2 * 3 * 4 = 24
```

An **iterable** is an object that can be traversed with `for x in ...`, such as a `list`, `tuple`, `dict`, `str`, or `set`.

An **iterator** is obtained with `iter()`, and `next()` retrieves its items one at a time. An iterator is exhausted after its items have been consumed.


```python
'''
next(iterator[, default])
iterator: an iterator object
default (optional): the value returned when the iterator is exhausted
If default is omitted, StopIteration is raised when the iterator is exhausted.
''';
```

```python
my_list = [1, 2, 3]

for item in iter(my_list):
    print(item, end = ' ')
```

```python
my_list = [1, 2, 3]
my_iter = iter(my_list)  # Create an iterator
# next(my_iter) returns an individual item, not an iterable
print(next(my_iter), type(next(my_iter)))
```

### Iteration


```python
# Traverse a list with a while loop
def while_traversal(list_traversal):
    """
    Traverse the elements of a list with a while loop.
    :param list_traversal: the list to traverse
    :return: None
    """
    ind=0  # Initialize the index
    while ind<len(list_traversal):
        print(list_traversal[ind])
        ind+=1

while_traversal([1,3,5,7])
```

```python
def for_traversal(list_traversal):
    """
    Traverse the elements of a list with a for loop.
    :param list_traversal: the list to traverse
    :return: None
    """
    for ele in list_traversal:
        print(ele)

for_traversal([0,2,4,6])
```

```python
def word_frequency(text: str):
    text = text.lower()
    for char in text:
        if not char.isalnum() and char != ' ':
            # isalnum() is true when all characters are alphanumeric and the string is non-empty
            text = text.replace(char, ' ')

    words = text.split()
    frequency_dict = {}

    for word in words:
        if word in frequency_dict:
            frequency_dict[word] += 1
        else:
            frequency_dict[word] = 1

    return frequency_dict

text = "Hello world! Hello NLP. World is fascinating."
frequency_dict = word_frequency(text)
frequency_dict
```

```python
def build_bow(corpus: list):
    """
    Build a Bag-of-Words representation.
    :param corpus: a list of sentences
    :return: a list containing the vocabulary and word-frequency vectors
    """
    # Build the vocabulary in order of first appearance
    vocab = []
    for sentence in corpus:
        words = sentence.lower().split()
        for word in words:
            if word not in vocab:
                vocab.append(word)

    # Generate a word-frequency vector for each sentence
    vectors = []
    for sentence in corpus:
        # Initialize a zero vector
        vector = [0] * len(vocab)
        words = sentence.lower().split()
        for word in words:
            # Find the word's position in the vocabulary
            index = vocab.index(word)
            # Increment the corresponding count
            vector[index] += 1
        vectors.append(vector)

    return [vocab] + vectors

corpus = ["NLP is fun", "Learning NLP and deep learning", "Deep learning is the future"]
result = build_bow(corpus)
result
```

```python
'''
For integers, & and | are bitwise AND and OR operators.
For sets, & and | compute intersection and union.
''';

def find_intersection_and_union(set1, set2):
    intersection = set1 & set2
    union = set1 | set2
    # intersection = set1.intersection(set2)
    # union = set1.union(set2)
    return intersection, union

set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
intersection, union = find_intersection_and_union(set1, set2)
intersection, union
```

```python
# Pitfall: modifying a list while iterating over it
nums = [1, 2, 3, 4]
for idx, n in enumerate(nums):
    if n % 2 == 0:
        nums.remove(n)  # An element may be skipped
        print(idx, nums)
```

```python
nums = [1, 2, 3, 4]
for idx, n in enumerate(nums[:]):  # Iterate over a copy
    if n % 2 == 0:
        nums.remove(n)  # Remove from the original list
        print(idx, nums)
```

```python
nums = [1, 2, 3, 4]
for idx in range(len(nums)-1, -1, -1):  # Iterate from the end to the beginning
    n = nums[idx]
    if n % 2 == 0:
        removed_n = nums.pop(idx)  # Remove the item at the current index
        print(idx, nums)
```

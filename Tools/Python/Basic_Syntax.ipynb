## Values and Variables


```python
# Literals are notations for constant values written directly in source code
print("Hello, Python!")
print(42, 3.14, True, None)
```

```python
# Assignment binds names to objects
width = 8
height = 5
area = width * height
print(area)

# Chained assignment binds several names to the same object
x = y = z = 10
print(x, y, z)
```

```python
# Every object has a type
data = "Python"
count = 3
ratio = 0.5

print(type(data))
print(type(count))
print(type(ratio))
print(type(None))
```

Names may contain Unicode letters, digits, and underscores, but a name cannot start with a digit. Names are case-sensitive and cannot be Python keywords. By convention, variable and function names use `lowercase_with_underscores`.


```python
# Common explicit conversions
print(int("42"), type(int("42")))
print(float("3.5"), type(float("3.5")))
print(str(42), type(str(42)))
print(bool(1), bool(0))

# Unicode code points and bytes
print(chr(65), ord("A"))
print("A".encode("utf-8"), b"A".decode("utf-8"))
```

```python
# int() converts a floating-point number by truncating toward zero
print(int(3.9))
print(int(-3.9))

# The optional base applies when converting a string
print(int("1010", 2))
print(int("ff", 16))
```

## Operators


```python
a = 7
b = 2

print(a + b)   # Addition
print(a - b)   # Subtraction
print(a * b)   # Multiplication
print(a / b)   # True division
print(a // b)  # Floor division
print(a % b)   # Remainder
print(a ** b)  # Exponentiation
```

```python
value = 5
value += 2
value *= 3
print(value)

# Exponentiation groups from right to left
print(2 ** 3 ** 2)
print(2 ** (3 ** 2))
```

```python
x = 5

print(1 < x < 10)          # Chained comparison
print(x == 5, x != 5)
print(x > 0 and x < 10)
print(x < 0 or x > 4)
print(not x == 5)
```

## Loops


```python
# A while loop repeats while its condition is true
i = 1
total = 0

while i <= 100:
    total += i
    i += 1

print(total)
```

```python
# A for loop iterates over the items of an iterable
word = "banana"
count = 0

for character in word:
    if character == "a":
        count += 1

print(count)
```

```python
print(list(range(5)))        # 0 through 4
print(list(range(2, 6)))     # 2 through 5
print(list(range(1, 8, 2)))  # Step of 2
```

```python
# continue skips the rest of the current iteration
# break exits the nearest enclosing loop
for number in range(2, 10):
    if number % 2 == 0:
        continue
    if number > 7:
        break
    print(number)
```

```python
# A loop's else clause runs when the loop finishes without break
value = 29

for divisor in range(2, value):
    if value % divisor == 0:
        print(f"{value} is composite")
        break
else:
    print(f"{value} is prime")
```

```python
# pass is a null statement used where syntax requires a statement
ready = False

if not ready:
    pass

print("Program continues")
```

```python
# Nested while loops: a 9-by-9 multiplication table
row = 1
while row < 10:
    column = 1
    while column <= row:
        print(f"{column}*{row}={column * row}", end=" ")
        column += 1
    row += 1
    print()
```

```python
# Nested for loops: a 9-by-9 multiplication table
for row in range(1, 10):
    for column in range(1, row + 1):
        print(f"{column}*{row}={column * row}", end=" ")
    print()
```

```python
def newton_sqrt(S, tolerance=1e-4, max_iterations=100):
    x_n = S / 2  # Initial guess
    for _ in range(max_iterations):
        x_next = 0.5 * (x_n + S / x_n)  # Update the estimate
        if abs(x_next - x_n) < tolerance:  # Check convergence
            return x_next
        x_n = x_next
    return x_n

S = 2
sqrt_S = newton_sqrt(S)
print(f"The square root of {S} is: {sqrt_S:.25f}")
```

## Functions

```python
'''
Functions
A function groups reusable statements under a name.

Function definition
def function_name(parameters):
    function_body
    return value

return leaves the current function call and optionally returns a value.

Function call
function_name(arguments)

Function encapsulation
Group statements that perform a related task in a separate function.

Function generalization
Use parameters so the same function can operate on different argument values.
''';
```

```python
def minus(x, y):
    """
    Return the difference between x and y
    :param x: minuend
    :param y: subtrahend
    :return: difference
    """
    # A string literal as the first statement of a function body is its docstring
    z = x - y
    return z


print(f"5-2={minus(5, 2)}")
```

```python
def hi():
    print("Hello World")


result = hi()
print(f"A function without an explicit return returns {result}, whose type is {type(result)}")
# None is commonly used to represent the absence of a value
# None is false in a Boolean context
while not result:
    print("None is false in a Boolean context")
    break
```

```python
def factor(n):
    """
    Return n factorial, written n!
    :param n: non-negative integer
    :return: n factorial
    """
    if not isinstance(n, int):
        raise TypeError('n must be an integer')
    if n < 0:
        raise ValueError('n must be non-negative')
    if n < 2:
        return 1
    return n * factor(n - 1)


print(factor(5))
```

```python
# A name assigned inside a function is local unless declared global or nonlocal
num = 2


def local_eg():
    num = 1
    print(num)


def global_eg():
    global num
    num = 3  # Global variable
    print(num)


local_eg()
global_eg()
print(num)
```

```python
def muti_return(x, y, z):
    f1 = x + y + z
    f2 = "xyz"
    f3 = x > y
    return f1, f2, f3  # Comma-separated return values are packed into a tuple


a0, b0, c0 = muti_return(1, 2, 3)
print(a0, b0, c0)

a1 = muti_return(1, 2, 3)[0]  # Indexing
print(a1)
```

```python
# Positional arguments
# Positional arguments are assigned to parameters in order
def user_info1(name, age, gender):
    print(f"Name: {name}, age: {age}, gender: {gender}")


user_info1("Alice", 20, "female")
```

```python
# Keyword arguments use the name=value form
# Positional arguments must precede keyword arguments in a call
# Keyword arguments can be supplied in any order
user_info1("Liz", age=21, gender="female")
```

```python
# Default argument values
def user_info2(name, age, gender="male"):  # Parameters with defaults follow required positional-or-keyword parameters
    print(f"Name: {name}, age: {age}, gender: {gender}")


user_info2("Tim", 22)
user_info2("Emmy", 20, "female")  # Supply a value instead of using the default
```

```python
# Variable-length positional arguments
def uncertain1(*args):  # args is a tuple
    return args


print(uncertain1("Mess"), type(uncertain1("Mess")))
print(uncertain1("Mess", 35), type(uncertain1("Mess", 35)))
```

```python
# Variable-length keyword arguments
def uncertain2(**kwargs):  # kwargs is a dictionary
    return kwargs


# Arguments use the key=value form
print(
    uncertain2(name="Mess", age=35, gender="male"),
    type(uncertain2(name="Mess", age=35, gender="male")),
)
```

```python
# Passing a function as an argument
def func12(compute):
    result = compute(1, 2)  # The data is fixed; the operation is passed in
    print(f"Argument type: {type(compute)}")
    print(f"func12 result: {result}")


def add(x, y):
    return x + y


# Function objects can be passed as arguments
func12(add)
```

```python
# Lambda expression
# Lambda expressions create anonymous functions restricted to a single expression
# lambda parameters: expression
func12(lambda x, y: x + y)
```

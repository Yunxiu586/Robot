```python
'''
Classes

class ClassName:
    class attributes
    methods

A method is a function associated with an object through its class.

def method_name(self, parameter1, parameter2, ...):
    method body

self is the conventional name for the instance parameter.
When a method is called through an instance, that instance is supplied as the first argument.
Instance attributes are commonly accessed through self inside methods.
''';
```

```python
class Info:
    def __init__(self):
        self.name = None
        self.age = None

    def prompt_msg(self):
        print('Enter a name and age')

    def info_print(self):
        print(f'Name: {self.name}\nAge: {self.age}')

    def msg_print(self, msg):
        print(f'{msg}')  # When called through an instance, self is supplied automatically
```

```python
# Assign instance attributes
student_1.name = 'Alice'
student_1.age = 20
```

```python
student_1.prompt_msg()
student_1.info_print()
student_1.msg_print('This is a message')
```

```python
class Directory2:

    def __init__(self, name, age, phone):
        self.name = name
        self.age = age
        self.phone = phone
        print(f'{self.name} is {self.age} years old; phone number: {self.phone}')

    # String representation
    def __str__(self):
        return f'Directory2(name={self.name}, age={self.age}, phone={self.phone})'

    # Less-than comparison
    def __lt__(self, other):
        return self.age < other.age

    # Less-than-or-equal comparison
    def __le__(self, other):
        return self.age <= other.age

    # Equality comparison
    def __eq__(self, other):
        return self.age == other.age
```

```python
contact2 = Directory2('Alice', 18, 987654321)
```

```python
print(contact2)
```

```python
print(str(contact2))
```

```python
contact3 = Directory2('Bob', 8, 24680246)
```

```python
print(f'Alice is older than Bob: {contact2 > contact3}')
print(f'Alice is younger than Bob: {contact2 < contact3}')
```

```python
print(f'Alice is at least as old as Bob: {contact2 >= contact3}')
print(f'Alice is at most as old as Bob: {contact2 <= contact3}')
```

```python
print(f'Alice and Bob have the same age: {contact2 == contact3}')
```

```python
'''
Object-oriented programming concepts

Encapsulation, inheritance, and polymorphism

1. Encapsulation

Attributes and behavior are grouped in a class as data attributes and methods.

Non-public members

Some attributes and methods are intended for internal use and are not accessed directly through the usual public interface.
''';
```

Uses of underscores

* `_name` marks a non-public name and is not imported by `from module import *`
* `name_` avoids conflicts with Python keywords
* `__name` triggers name mangling for a class attribute
* `__name__` uses leading and trailing double underscores for special names defined by Python, such as `__init__` and `__str__`

```python
class Phone1:

    __current_voltage = 1.5  # Name-mangled data attribute
    producer = 'Apple'

    def __keep_single_core(self):  # Name-mangled method
        print('Phone1 is running with one CPU core')

    def call_by_5G(self):
        if self.__current_voltage >= 1:  # Access a name-mangled data attribute
            print('Phone1 5G calling is enabled')
        else:
            self.__keep_single_core()  # Access a name-mangled method
            print('Phone1 has insufficient power for 5G calling and switched to one CPU core')
```

```python
phone1 = Phone1()
phone1.call_by_5G()
```

```python
# Direct access fails because the name is mangled
try:
    print(Phone1.__current_voltage)
except Exception as error:
    print(error)
```

```python
print(Phone1._Phone1__current_voltage)
```

```python
Phone1._Phone1__current_voltage = 2
print(Phone1._Phone1__current_voltage)
```

```python
class Parent:
    _num = 10  # Inherited and accessible

    def print_num(self):
        print(self._num)

class Child(Parent):
    def print_num(self):
        print(self._num)

parent = Parent()
child = Child()

parent.print_num()  # 10
child.print_num()   # 10
```

```python
class Parent:
    __num = 10

    def print_num(self):
        print(f'Parent: {self.__num}')

class Child(Parent):
    def print_num(self):
        try:
            print(self.__num)
        except Exception as error:
            self.__num = 20
            print(error)
            print(f'Child: {self.__num}')

parent = Parent()
child = Child()

parent.print_num()  # 10
child.print_num()   # 20
```

```python
# Override attributes and methods inherited from a base class
class Phone3(Phone1):
    producer = 'Huawei'  # Override a base-class attribute

    def call_by_5G(self):  # Override a base-class method
        print('Phone3 enabled one-core CPU mode')
        print('Phone3 is making a call over the 5G network')

        # Access a base-class attribute
        print(f'Base-class attribute: {Phone1.producer}')
        print(f'Base-class attribute: {super().producer}')

        # Call a base-class method
        Phone1.call_by_5G(self)
        super().call_by_5G()
```

```python
phone3 = Phone3()
```

```python
print(f'phone1 producer: {phone1.producer}')
```

```python
print(f'phone3 producer: {phone3.producer}')  # Access the overridden attribute
```

```python
phone3.call_by_5G()  # Call the overridden method
```

```python
'''
3. Polymorphism

The same operation can produce different behavior for different objects.
A function can operate on any object that provides the required methods; inheritance is not required.

An abstract base class defines a common interface with abstract methods.
A class with abstract methods cannot be instantiated until all abstract methods and properties are overridden.
''';
```

```python
from abc import ABC, abstractmethod


class Animal(ABC):
    @abstractmethod
    def speak(self):
        """Produce the animal's sound""";
        pass

    @abstractmethod
    def move(self):
        """Perform the animal's movement""";
        pass
```

```python
class Dog(Animal):
    def speak(self):
        print('Woof')

    def move(self):
        print('Run')
```

```python
class Bird(Animal):
    def speak(self):
        print('Chirp')

    def move(self):
        print('Fly')
```

```python
def make_noise(animal: Animal):
    animal.speak()
```

```python
dog = Dog()
bird = Bird()
```

```python
make_noise(dog)
make_noise(bird)
```

```python
dog.move()
bird.move()
```

# imports

A module is a file containing Python definitions and statements. An import statement makes names defined by a module available to the program.

```python
# Import a module and access its names with dot notation
import math

math.sqrt(16)
```

```python
# Import a selected name directly
from math import pi

pi
```

```python
# Bind a module to an alias
import math as m

m.factorial(5)
```

```python
# Bind an imported name to an alias
from math import sqrt as square_root

square_root(25)
```

`from module import *` imports the module's public names directly. When imports bind the same name, the later binding replaces the earlier one.

### custom modules

A custom module is a `.py` file containing definitions and statements. Place it where Python can find it, then import it using the file name without the `.py` suffix. Use `module.name` after importing the module, or import selected names directly. Aliases can be used with either form.

### packages

A regular package organizes modules in a directory and usually contains an `__init__.py` file. Packages provide dotted module names such as `package.module`. Common forms include `import package.module`, `from package import module`, and `from package.module import name`.

# math

```python
import math

# Mathematical constants
print(f"Euler's number e: {math.e}")
print(f"Pi: {math.pi}")
print(f"Infinity: {math.inf}")
print(f"Not a number: {math.nan}")
```

```python
x = 2.25
y = 4.9
print(f"ceil({x}) = {math.ceil(x)}")  # Smallest integer greater than or equal to x
print(f"floor({x}) = {math.floor(x)}")  # Largest integer less than or equal to x
print(f"trunc({x}) = {math.trunc(x)}")  # Integer part of x, rounded toward zero
print(f"fabs({x}) = {math.fabs(x)}")  # Absolute value
print(f"sqrt({x}) = {math.sqrt(x)}")  # Square root
print(f"pow({x}, {y}) = {math.pow(x, y)}")  # Raise x to the power y
print(f"exp(2) = {math.exp(2)}")  # Raise e to the given power
print(f"ln(e) = {math.log(math.e)}")  # Natural logarithm
print(f"log10(100) = {math.log10(100)}")  # Base-10 logarithm
```

```python
# Trigonometric functions use radians
angle_deg = 45
angle_rad = math.radians(angle_deg)  # Convert degrees to radians
angle_rad = angle_deg * math.pi / 180
print(f"{angle_deg}° = {angle_rad} radians")
print(f"sin({angle_rad}) = {math.sin(angle_rad)}")
print(f"cos({angle_rad}) = {math.cos(angle_rad)}")
print(f"tan({angle_rad}) = {math.tan(angle_rad)}")
```

```python
# Other functions
print(f"gcd(12, 18) = {math.gcd(12, 18)}")  # Greatest common divisor
print(f"factorial(5) = {math.factorial(5)}")  # Factorial
```

# random

```python
import random

# Return a random floating-point number in [0.0, 1.0)
random.random()
```

```python
# uniform(a, b) returns N such that a <= N <= b when a <= b
# The endpoint b may or may not be included due to floating-point rounding
random.uniform(5, 10)
```

```python
# randint(a, b) returns a random integer in [a, b]
random.randint(1, 6)
```

```python
# choice(seq) returns a randomly selected element from a non-empty sequence
fruits = ["apple", "banana", "orange", "grape", "strawberry"]
random.choice(fruits)
```

```python
# shuffle(seq) shuffles a mutable sequence in place
cards = ["Ace of Hearts", "King of Spades", "Queen of Diamonds", "Jack of Clubs", "Ten of Hearts"]
random.shuffle(cards)
print(cards)
```

```python
# sample(population, k) performs random sampling without replacement
numbers = list(range(1, 11))
selected = random.sample(numbers, 3)
print(selected)
```

```python
# randrange(start, stop, step) returns a randomly selected element from range(start, stop, step)
random.randrange(0, 11, 2)
```

```python
# Initialize the random number generator's internal state
random.seed(42)
print(random.randint(1, 100), random.randint(1, 100))
random.seed(42)
print(random.randint(1, 100), random.randint(1, 100))
```

# copy

```python
import copy

# Original list
a = [1, 2]
b = copy.copy(a)      # Shallow copy
c = copy.deepcopy(a)  # Deep copy

# Modify the original list
a[1] = [1, 2]
print("Original:", a)
print("Shallow copy:", b)
print("Deep copy:", c)
```

```python
a = [3, 4]
b = copy.copy(a)      # Shallow copy
c = copy.deepcopy(a)  # Deep copy
b[1] = [3, 4]
print("Original:", a)
print("Shallow copy:", b)
print("Deep copy:", c)
```

```python
a = [5, 6]
b = copy.copy(a)      # Shallow copy
c = copy.deepcopy(a)  # Deep copy
c[1] = [5, 6]
print("Original:", a)
print("Shallow copy:", b)
print("Deep copy:", c)
```

```python
a = [1, [0, 0]]
b = copy.copy(a)      # Shallow copy
c = copy.deepcopy(a)  # Deep copy
a[1][1] = 2           # Modify a nested object shared with the shallow copy
print("Original:", a)
print("Shallow copy:", b)
print("Deep copy:", c)
```

```python
a = [3, [0, 0]]
b = copy.copy(a)      # Shallow copy
c = copy.deepcopy(a)  # Deep copy
b[1][1] = [4, 4]      # Modify a nested object shared with the original
print("Original:", a)
print("Shallow copy:", b)
print("Deep copy:", c)
```

```python
a = [5, [0, 0]]
b = copy.copy(a)      # Shallow copy
c = copy.deepcopy(a)  # Deep copy
c[1][1] = [6, 6]
print("Original:", a)
print("Shallow copy:", b)
print("Deep copy:", c)
```

```python
# Handle an exception
try:
    result = 1 / 0
    print(result, type(result))
except:
    result = float("inf")
    print(result, type(result))
```

```python
# Handle a specific exception
try:
    print(undefined_name)
except NameError as error:
    print("The variable is not defined")
    print(error, type(error))
```

```python
# Handle multiple exceptions with one except clause
try:
    print(undefined_name)
    result = 1 / 0
except (NameError, ZeroDivisionError) as error:
    print("The variable is not defined or division by zero occurred")
    print(error, type(error))
```

```python
# Handle any exception derived from Exception
try:
    number = int("not a number")
except Exception as error:
    print(f"An exception occurred: {error}")
    print(type(error))
```

```python
# The else clause runs if the try clause completes without raising an exception
# The finally clause executes whether or not an exception occurs
try:
    print("Hello, world!")
except Exception as error:
    print(f"An exception occurred: {error}")
else:
    print("No exception occurred")
finally:
    print("This block always runs")
```

```python
# An unhandled exception in a function propagates to its caller
def divide_by_zero():
    return 1 / 0

def call_function():
    try:
        divide_by_zero()
    except Exception as error:
        print(f"The called function raised an exception: {type(error)}")

call_function()
```
